## Cell 3: Download & Prepare Drone Footage (Mill-19 Building)

```python
import os
from pathlib import Path

DATA_DIR = Path("./data/mill_drone")
OUTPUT_DIR = Path("./output/splat_model")
DATA_DIR.mkdir(parents=True, exist_ok=True)
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

# Download sample drone imagery with camera poses
# If using your custom drone photos, you can upload a zip file directly to Colab
print(f"Dataset Directory: {DATA_DIR.resolve()}")
print(f"Output Directory : {OUTPUT_DIR.resolve()}")
```

## Cell 4: Run 3D Gaussian Splatting Optimization (7,000 steps, ~25 min)

```python
# Train 3DGS model using gsplat
!python -m gsplat.examples.simple_trainer \
    --data_dir {DATA_DIR} \
    --result_dir {OUTPUT_DIR} \
    --max_steps 7000 \
    --eval_steps 1000 \
    --save_steps 7000

print("[SUCCESS] 3DGS Optimization Complete!")
```

## Cell 5: Download point_cloud.ply to Your Local Project

```python
from google.colab import files
import glob

ply_candidates = glob.glob(f"{OUTPUT_DIR}/**/point_cloud.ply", recursive=True)
if ply_candidates:
    target_ply = ply_candidates[0]
    print(f"Found point cloud: {target_ply} ({round(os.path.getsize(target_ply)/(1024*1024), 2)} MB)")
    files.download(target_ply)
    print("Download started! Place this file into: spatial-twin/output/mill/point_cloud.ply")
else:
    print("[WARN] point_cloud.ply not found in output directory.")
```
